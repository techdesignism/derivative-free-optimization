```python
import nevergrad as ng
from obj_funcs_lib import *
import numpy as np
import time

def optimize_with_bo(start_point, objective, bo_budget=100):
    """
    Optimizes a given `objective` function starting from `start_point`
    using Bayesian Optimization from nevergrad.
    """
    bo_optimizer_cls = ng.optimizers.registry['BO']
    parametrization = ng.p.Array(init=start_point)
    bo_optimizer = bo_optimizer_cls(parametrization=parametrization, budget=bo_budget)
    bo_result = bo_optimizer.minimize(lambda x: objective(x))
    current_pos = np.array(bo_result.value)
    current_obj_value = objective(current_pos)
    print(current_pos)
    print(current_obj_value)
    return current_pos, current_obj_value

results = {}
for name, obj_func in OBJ_FUNCS.items():
    print(f"Optimizing {name} with BO...")
    start_time = time.time()
    result = optimize_with_bo(
        start_point=[200]*3,
        objective=obj_func,
    )
    elapsed = time.time() - start_time
    print(f"Time taken for {name}: {elapsed:.2f} seconds")
    results[name] = result
    print('='*20)
```
