```python
#P-GCS
from numpy import *
import matplotlib.pyplot as plt
from obj_funcs_lib import *

from scipy.optimize import minimize

import numpy as np
np.set_printoptions(precision=2)
import datetime
import nevergrad as ng

import concurrent.futures


def t(x):
    return np.square( np.tan(np.pi*x-np.pi/2) )
def T(x):
    return (np.pi/2-np.arctan( np.sqrt(x) ) )/np.pi
def s(x,a,b):
    return ((np.cos(2*np.pi*x-np.pi)+1)/2)*(T(b)-T(a))+T(a)
def w(x,a,b):
    return t(s(x,a,b))


def optimize_with_pgcs(
    start_point, 
    objective,
    target=0.05, 
    w_a=0, 
    w_b=np.inf, 
    crunch_period=5000, 
    N=700000
):
    """
    Optimizes a given `objective` function starting from `start_point` with options for
    target value, step size parameters w_a, w_b, crunching period, and the number of iterations N.
    """
    sd = w_a
    current_pos = np.array(start_point)
    current_obj_value = objective(current_pos)
    crunch_step = 0
    w_cache = np.zeros(crunch_period)

    # Precompute w_cache
    for i in range(crunch_period):
        w_cache[i] = w(i / crunch_period, w_a, w_b)

    i = 0
    while i < N:
        if i == 0:
            #powell
            powell_result = minimize(objective, current_pos, method='Powell')
            current_pos = powell_result.x
            current_obj_value = powell_result.fun

        #GCS explode
        temp_pos = current_pos + np.random.normal(0, sd, np.array(start_point).shape)
        temp_obj_value = objective(temp_pos)
        if temp_obj_value < current_obj_value:
            temp_powell = minimize(objective, temp_pos, method='Powell')
            temp_powell_pos = temp_powell.x
            temp_powell_value = temp_powell.fun

            if temp_powell_value < temp_obj_value:
                temp_obj_value = temp_powell_value
                temp_pos = temp_powell_pos
            current_pos = temp_pos
            current_obj_value = temp_obj_value

            rand_int = np.random.randint(0, 100)
            if rand_int <= 1:
                print('TinySQP')
                optimizer_cls = ng.optimizers.registry['TinySQP']
                optimizer = optimizer_cls(parametrization=ng.p.Array(init=start_point), budget=500)
                temp_tinysqp_pos = optimizer.minimize(lambda x: objective(x))
                temp_tinysqp_value = objective(temp_tinysqp_pos.value)
                if temp_tinysqp_value < current_obj_value:
                    current_pos = temp_tinysqp_pos.value
                    current_obj_value = temp_tinysqp_value

        else:
            crunch_step += 1
        if current_obj_value < target:
            break
        if crunch_step % crunch_period == 0:
            crunch_step = 0
        sd = w_cache[crunch_step]
        i += 1

    print(current_pos)
    print(objective(current_pos))
    return current_pos, objective(current_pos)



import time

results = {}
for name, obj_func in OBJ_FUNCS.items():
    print(f"Optimizing {name}...")
    start_time = time.time()
    result = optimize_with_pgcs(
        start_point=[200]*3,
        objective=obj_func,
    )
    elapsed = time.time() - start_time
    print(f"Time taken for {name}: {elapsed:.2f} seconds")
    results[name] = result
    print('='*20)


```

```python
from obj_funcs_lib import *
import nevergrad as ng
import numpy as np
import time

# Only use TinySQP optimizer
results = {}
for name, obj_func in OBJ_FUNCS.items():
    print(f"Optimizing {name}...")
    start_time = time.time()
    optimizer_cls = ng.optimizers.registry['TinySQP']
    optimizer = optimizer_cls(parametrization=ng.p.Array(init=[200.0]*3), budget=500)
    recommendation = optimizer.minimize(lambda x: obj_func(x))
    value = obj_func(recommendation.value)
    sci_fmt = {"float_kind": lambda x: np.format_float_scientific(x, precision=2, unique=False)}
    print(np.array2string(np.asarray(recommendation.value), formatter=sci_fmt))
    print(np.array2string(np.asarray([value]), formatter=sci_fmt))
    elapsed = time.time() - start_time
    print(f"Time taken for {name}: {elapsed:.2f} seconds")
    print('='*20)
    results[name] = (recommendation.value, value)
```
